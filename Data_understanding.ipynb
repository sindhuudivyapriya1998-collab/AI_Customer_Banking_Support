# Data quality analysis

```python
train['text_length'] = train['text'].str.len()
train['word_count'] = train['text'].str.split().str.len()

print(train['text_length'].describe())
print(train['word_count'].describe())

print("empty texts:",(train['text'].str.strip()== '').sum())
print("very short (<3 words):",(train['word_count'] < 3).sum())

print(train['text'].str.contains(r'[^\x00-\x7F]').sum(),"non-ascii rows")
```
